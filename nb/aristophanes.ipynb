```python
from pathlib import Path
import random
import re
from typing import Iterator

ROOT = Path.cwd().parent

# load the 2, 3, and 4 strophic baselines separately
tetrameter_2 = ROOT / "data/compiled/baselines/baseline_aristophanes/tetrameter_2.xml"
tetrameter_3 = ROOT / "data/compiled/baselines/baseline_aristophanes/tetrameter_3.xml"
tetrameter_4 = ROOT / "data/compiled/baselines/baseline_aristophanes/tetrameter_4.xml"

trimeter_2 = ROOT / "data/compiled/baselines/baseline_aristophanes/trimeter_2.xml"
trimeter_3 = ROOT / "data/compiled/baselines/baseline_aristophanes/trimeter_3.xml"
trimeter_4 = ROOT / "data/compiled/baselines/baseline_aristophanes/trimeter_4.xml"

# make permutated versions of all canticum elements in each of the six files.
# - if we can hold the versions in memory, that is best, otherwise we have to save 60 000 files to disk
# - what's permuted are the l element orderings of each of its strophes.
# - it is key that the permutation is different for all strophes inside the same canticum, so responsion is not conserved, otherwise the permutation test will not work
# - there are n! possible permutations for a strophe with n lines, which is why we need to go Monte Carlo

permutations = 10_000

temp_dir = ROOT / "data/compiled/baselines/baseline_aristophanes/temp_permutations"

baseline_files = [
    tetrameter_2,
    tetrameter_3,
    tetrameter_4,
    trimeter_2,
    trimeter_3,
    trimeter_4,
]

strophe_re = re.compile(r"(<strophe\b[^>]*>)(.*?)(</strophe>)", re.S)


def shuffle_strophe_lines(strophe_body: str, rng: random.Random) -> str:
    lines = strophe_body.splitlines(keepends=True)

    l_indices = [i for i, line in enumerate(lines) if re.search(r"<l\b", line)]

    if len(l_indices) <= 1:
        return strophe_body

    l_lines = [lines[i] for i in l_indices]
    rng.shuffle(l_lines)

    for idx, new_line in zip(l_indices, l_lines):
        lines[idx] = new_line

    return "".join(lines)


def permute_text_once(text: str, rng: random.Random) -> str:

    def repl(match):
        open_tag, body, close_tag = match.groups()
        new_body = shuffle_strophe_lines(body, rng)
        return open_tag + new_body + close_tag

    return strophe_re.sub(repl, text)


def iter_permuted_texts(
    xml_path: Path,
    n_permutations: int,
    seed: int | None = None,
) -> Iterator[str]:

    text = xml_path.read_text(encoding="utf-8")

    rng = random.Random(seed)

    for _ in range(n_permutations):
        yield permute_text_once(text, rng)


def save_permutations(
    xml_path: Path,
    n_permutations: int,
    output_dir: Path,
    seed: int | None = None,
) -> None:

    output_dir.mkdir(parents=True, exist_ok=True)
    stem = xml_path.stem

    for i, permuted_text in enumerate(
        iter_permuted_texts(xml_path, n_permutations, seed=seed),
        start=1,
    ):
        out_path = output_dir / f"{stem}_perm_{i:05d}.xml"
        out_path.write_text(permuted_text, encoding="utf-8")


def print_test_permutation(xml_path: Path) -> None:
    text = next(iter_permuted_texts(xml_path, 1, seed=42))
    print(text[:10000])
```

```python
# Test print of one permutation to inspect
print_test_permutation(tetrameter_3)
```

```python
for file in baseline_files:
    save_permutations(
        file,
        n_permutations=permutations,
        output_dir=temp_dir / file.stem,
        seed=1453,
    )
```
