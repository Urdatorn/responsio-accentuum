```python
from pathlib import Path

def find_project_root(start: Path, markers=("pyproject.toml", ".git")):
    for p in [start] + list(start.parents):
        if any((p / m).exists() for m in markers):
            return p
    raise RuntimeError("Project root not found")

ROOT = find_project_root(Path.cwd())
```

Here we compute the contours and scores for the three first and final four lines of the famous entry of the Clouds that is used as example in the proceedings paper.

```python
from pathlib import Path
ROOT = Path.cwd().parent

from responsio_accentuum import compatibility_canticum, compatibility_ratios_to_stats, flatten_recursive

fractions = compatibility_canticum(ROOT / "data/compiled/extra/test/test_nu01.xml", "nu01", print_contours=True)
score = compatibility_ratios_to_stats(fractions)

print("Normalized score per syll:")
print(" ".join(str(frac) for frac in flatten_recursive(fractions)))

print(f"\nMean song score: {float(score)} (= {score})")
```
